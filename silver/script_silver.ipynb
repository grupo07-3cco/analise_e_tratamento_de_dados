# Tratamento da camada Silver

Este notebook executa a limpeza e a padronizacao das leituras presentes em `bronze/dados`. O processamento remove registros invalidos ou duplicados exatos, converte datas e valores para tipos consistentes, e grava arquivos individuais por sensor e um arquivo consolidado em `silver`.

```python
from pathlib import Path
import sys

pasta_silver = Path.cwd()
if pasta_silver.name != "silver":
    pasta_silver = pasta_silver / "silver"
sys.path.insert(0, str(pasta_silver))

from processar_sensores import processar_dados

leituras = processar_dados()
leituras.head()
```

```python
leituras.groupby(["sensor", "unidade"], dropna=False).agg(
    quantidade=("valor", "size"),
    minimo=("valor", "min"),
    maximo=("valor", "max"),
).reset_index()
```
